```python
import pandas as pd
import numpy as np

print("Pandas imported successfully!")
```

```python
data = {
    "Name": ["Aman", "Priya", "Rahul", "Sneha", "Aman", "Vikas", "Neha", "Riya"],
    "Age": [22, 25, np.nan, 24, 22, 28, 23, 26],
    "Salary": [25000, 32000, 28000, np.nan, 25000, 45000, 30000, 35000],
    "City": ["Pune", "Mumbai", "Pune", "Delhi", "Pune", "Mumbai", "Delhi", "Pune"]
}

df = pd.DataFrame(data)

df
```

```python
print("First 5 rows:")
display(df.head())

print("\nDataset Information:")
df.info()
```

```python
print("Missing values in each column:")
print(df.isnull().sum())
```

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Salary"] = df["Salary"].fillna(df["Salary"].median())

print("Missing values after cleaning:")
print(df.isnull().sum())
```

```python
df = df.drop_duplicates()

print("Duplicates after cleaning:")
print(df.duplicated().sum())
```

```python
print("Statistical Summary:")
df.describe()
```

```python
print("Number of people from each city:")
print(df["City"].value_counts())
```

```python
print("Final Cleaned Dataset:")
display(df)

print("Shape of dataset:", df.shape)
```

```python
print("Conclusion:")
print("The dataset was successfully cleaned and explored.")
print("Missing values were handled and duplicate records were removed.")
print("Basic statistical and categorical analysis was performed.")
```
