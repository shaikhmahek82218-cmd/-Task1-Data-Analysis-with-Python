```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
data = {
    "Name": ["Aman", "Priya", "Rahul", "Sneha", "Aman", "Vikas", "Neha", "Riya"],
    "Age": [22, 25, 24, 24, 22, 28, 23, 26],
    "Salary": [25000, 32000, 28000, 32000, 25000, 45000, 30000, 35000],
    "City": ["Pune", "Mumbai", "Pune", "Delhi", "Pune", "Mumbai", "Delhi", "Pune"]
}

df = pd.DataFrame(data)

df
```

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nStatistical Summary:")
display(df.describe())
```

```python
plt.figure(figsize=(7, 5))

plt.hist(df["Age"], bins=5)

plt.xlabel("Age")
plt.ylabel("Number of People")
plt.title("Age Distribution")

plt.show()
```

```python
city_counts = df["City"].value_counts()

plt.figure(figsize=(7, 5))
plt.bar(city_counts.index, city_counts.values)

plt.xlabel("City")
plt.ylabel("Number of People")
plt.title("Number of People by City")

plt.show()
```

```python
plt.figure(figsize=(7, 5))

plt.scatter(df["Age"], df["Salary"])

plt.xlabel("Age")
plt.ylabel("Salary")
plt.title("Age vs Salary")

plt.show()
```

```python
print("Highest Salary:")
print(df["Salary"].max())

print("\nLowest Salary:")
print(df["Salary"].min())

print("\nAverage Salary:")
print(df["Salary"].mean())

print("\nAverage Age:")
print(df["Age"].mean())
```

## Key Findings

1. The dataset contains information about age, salary, and city.
2. The age distribution shows that most people are in the younger age group.
3. Pune has the highest number of people in the dataset.
4. The salary values show variation among individuals.
5. The scatter plot helps to observe the relationship between age and salary.
6. No major unusual values were observed in this small sample dataset.
